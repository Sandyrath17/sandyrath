## Visualization Choice

The one visual that caught my curiosity from the *Directory of Visualizations* and which we didn't discuss in the class is the **violin plot**. It is a kind of plot that is usually used to display the distribution of a continuous variable using a density estimate. Unlike histogram, violin plot helps to show the full shape of the distribution which usually facilitates in seeing clustering data. Wilke explains in the *Directory of Visualizations* (Distributions section) that distribution plots help compare variation across groups by revealing the structure of the data rather than just summary values. 

In *Aesthetic Mapping*, Wilke discusses how effective visualizations depend on mapping data variables to perceptual features like position and size. In a violin plot, the numerical variable is usually mapped along the vertical axis while the width of the shape represents the density of observations at each value. The visual encoding allows viewers to interpret how values are distributed across a range (Wilke, *Aesthetic Mapping*, sections of position and size encodings).

In terms of my career goals in media and communication, a violin plot could be used in visualizing how public opinion varies across demographic groups. Rather than presenting only an average evaluation score, a violin plot would help in showing whether opinions are clustered within specific populations. This would provide deeper insight into the patterns of audience responses.

## Improvement for the ConMediaUse and ImmiPhobia scatterplot

I think the scatterplot of ConMediaUse and ImmiPhobia may have a hard time from overplotting if many data points start overlapping. In *Overlapping Points*, Wilke explains how overlapping points can indeed obscure patterns in dense datasets and make it quite difficult in terms of interpretation of data. A strategy that can be recommended for this would be the use of transparency. Through this strategy, the areas with many overlapping points will appear darker and reveal underlying structure (Wilke, *Overlapping Points*, section on transparency).

Also, I would like to add that Wilke discusses another significant practice in *Aesthetic Mapping* by emphasizing that visual properties like position, size, and opacity affect how clearly patterns are perceived. Adding transparency or adjusting marker size would improve readability without changing the data itself (Wilke, *Aesthetic Mapping*, sections on visual encoding). 

Through the practice of applying these above principles could help a lot in making the scatterplot clearer and even improve interpretation of the relationship between conservative media use and immigration attitutes.
