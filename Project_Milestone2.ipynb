## Milestone 2

In this milestone, I have decided to analyse the relationships between audio features and song popularity. I am focusing on features like energy and danceability to examine how they relate to popularity. I am also comparing these relationships in different genres to determine whether the patterns vary considering every genre. 

From Milestone 1, the observations that were noticed: the histogram showed the distribution of song popularity across the dataset and the bar chart represented the number of songs that each genre had in the dataset. The bar chart showed that the genres have technically similar number of songs which resulted to the bars of similar height. This indicates that the dataset is balanced across the dataset. This is important to observe for the analysis because it allows to fairly compare relationships between audio features and popularity in different genres. 

```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv("dataset.csv")
```

```python
df = df [['track_name', 'artists', 'track_genre', 'popularity', 'danceability', 'energy', 'valence', 'tempo']]
```

```python
df = df.dropna()
```

The initial scatter plots show the overall relationship between audio features and popularity across the entire dataset. These plots are helping in identifying genral trends but the relationships seem weak and widely spread. This suggests that popularity is not strongly determined by a single feature when all genres are combined. 

```python
fig, ax = plt.subplots()
ax.scatter(df['energy'], df['popularity'], alpha = 0.5)
ax.set_title("Energy vs Popularity")
ax.set_xlabel("Energy")
ax.set_ylabel("Popularity")
plt.show()
```

```python
fig, ax = plt.subplots()
ax.scatter(df['danceability'], df['popularity'], alpha = 0.5)
ax.set_title("Danceability vs Popularity")
ax.set_xlabel("Danceability")
ax.set_ylabel("Popularity")
plt.show()
```

In order to better understand these relationships, I decided to flter the dataset by genre to compare relationships between variables within specific categories. Since the dataset contains equal number of songs across genres, comparisons between genres are not biased by differences in sample size. This helps in having clearer analysis to understand how audio features may influence popularity within specific genres. 

```python
pop_df = df[df['track_genre'] == 'pop']
rock_df = df[df['track_genre'] == 'rock']
hiphop_df = df[df['track_genre'] == 'hip-hop']
```

Important note: to keep my analysis clear and focused, I only selected a few representative genres (pop, rock, and hip-hop) for comparisons. This helps in providing a more detailed analysis instead of overwhelming the analysis with too many categories. 

```python
fig, ax = plt.subplots()
ax.scatter(pop_df['energy'], pop_df['popularity'], alpha = 0.5)
ax.set_title("Energy vs. Popularity (Pop)")
ax.set_ylabel("Energy")
ax.set_xlabel("Popularity")
plt.show()
```

```python
fig, ax = plt.subplots()
ax.scatter(rock_df['energy'], rock_df['popularity'], alpha = 0.5)
ax.set_title("Energy vs. Popularity (Rock)")
ax.set_ylabel("Energy")
ax.set_xlabel("Popularity")
plt.show()
```

```python
fig, ax = plt.subplots()
ax.scatter(hiphop_df['energy'], hiphop_df['popularity'], alpha = 0.5)
ax.set_title("Energy vs. Popularity (Hip-Hop)")
ax.set_ylabel("Energy")
ax.set_xlabel("Popularity")
plt.show()
```

```python
fig, ax = plt.subplots()
ax.scatter(pop_df['danceability'], pop_df['popularity'], alpha = 0.5)
ax.set_title("Danceability vs. Popularity (Pop)")
ax.set_ylabel("Danceability")
ax.set_xlabel("Popularity")
plt.show()
```

When analysing the scatter plots by genre, differences in patterns appear. There are some genres that show clustering between features and popularity but others remain more scattered. This is suggesting that the relationship between audio features and popularity is not consistent across genres and might depend on the specific characteristics of each genre.
