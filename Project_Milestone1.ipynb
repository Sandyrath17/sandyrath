```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv("dataset.csv")
```

**Data Cleaning** 

After looking at a variety of projects done with the same dataset, I chose to refer to the following Kaggle project to guide my data cleaning process: https://www.kaggle.com/code/jalesummak/spotify-track-ranking-project#5.-Data-Cleaning

```python
df.columns
```

I selected a set of relevant variables: track name, artist, genre, popularity, and key audio features like danceability, energy, etc here.

```python
df = df[['track_name', 'artists', 'track_genre', 'popularity', 'danceability', 'energy', 'valence', 'tempo']]
```

I removed the missing values just like the linked reference did from the dataset so that there is consistency and no issues during the analysis. 

```python
df = df.dropna()
```

I checked the dimensions of the dataset after cleaning to confirm the number of observations being used. 

```python
df.shape
```

Just to be sure that there are no remaining missing values in the dataset after cleaning.

```python
df.isnull().sum()
```

End of reference

**Visualizations to understand the overall data**

**#1 Histogram**

To examine the distribution of song popularity across the dataset.

```python
fig, ax = plt.subplots()

ax.hist(df['popularity'])
ax.set_title(("Popularity Histogram"))

plt.show()
```

**#2 Bar Chart**

To explore how songs are distributed across different genres. 

```python
fig, ax = plt.subplots()

df['track_genre'].value_counts().plot.bar()

ax.set_title("Number of Songs by Genre")

plt.show()
```

The above bar graph is overwhelming to understand as it is showing the number of songs by all the genres present in the dataset. In order to make the information to be easily perceived, we can analyse the data by looking at 10 genres at a time.

```python
fig, ax = plt.subplots()

df['track_genre'].value_counts().head(10).plot.bar()

ax.set_title("Number of Songs by Genre")

plt.show()
```
